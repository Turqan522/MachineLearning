```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Alzheimer Disease and Healthy Aging Data In US.csv')
df
```

```python
df=df.drop(columns=['Sample_Size'])
```

```python
df.isna().sum()
```

```python
pd.set_option('display.max_columns', None)

display(df.head(100))
```

```python
df['Data_Value']=df['Data_Value'].fillna(df['Data_Value'].median())
```

```python
df.isna().sum()
```

```python
df['Data_Value_Alt']=df['Data_Value_Alt'].fillna(df['Data_Value_Alt'].median())
```

```python
df.info()
```

```python
df['Low_Confidence_Limit']
```

```python
df['Low_Confidence_Limit']=pd.to_numeric(df['Low_Confidence_Limit'],errors='coerce')
```

```python
df['Low_Confidence_Limit']=df['Low_Confidence_Limit'].fillna(df['Low_Confidence_Limit'].median())
```

```python
df['High_Confidence_Limit']=pd.to_numeric(df['High_Confidence_Limit'],errors='coerce')
df['High_Confidence_Limit']=df['High_Confidence_Limit'].fillna(df['High_Confidence_Limit'].median())
```

```python
df['StratificationCategory2'].head(5000)
```

```python
df['StratificationCategory2'].value_counts(dropna=False)
```

```python
df['StratificationCategory2']=df['StratificationCategory2'].fillna('Unknown')
```

```python
df['Stratification2'].value_counts(dropna=False)
```

```python
df['Stratification2']=df['Stratification2'].astype(str).str.replace('/',',')
```

```python
df['Stratification2']=df['Stratification2'].fillna('Unknown')
```

```python
df['Geolocation']=df['Geolocation'].fillna('Unknown')
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df
```

```python
df[['Data_Value', 'Data_Value_Alt']].head(10)
df[['Data_Value', 'Data_Value_Alt']].dtypes
```

```python
# fərqli olan sətirlərin sayı (NaN-ları eyni sayaraq)
diff = ~((df['Data_Value'] == df['Data_Value_Alt']) | (df['Data_Value'].isna() & df['Data_Value_Alt'].isna()))
print(diff.sum())
```

```python
df=df.drop(columns=['YearStart','YearEnd','LocationAbbr','LocationDesc','Data_Value_Alt','Geolocation','ClassID','TopicID','QuestionID','LocationID'])
```

```python
df
```

```python
df.groupby('Datasource')['Data_Value'].mean()
```

```python
df.groupby('Topic')['Low_Confidence_Limit'].agg(['median','mean'])
```

```python
df.groupby('Stratification1')['Data_Value'].agg(['median','mean'])
```

```python
df.groupby('Data_Value_Type')['Data_Value'].agg(['median','mean'])
```

```python
plt.figure(figsize=(10,6))
df.groupby('Topic')['Low_Confidence_Limit'].agg(['median','mean']).plot(kind='bar')
plt.xlabel('Topic')
plt.ylabel('Low Confidence Limit')
plt.title('Low Confidence Limit by Topic')
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby('Stratification1')['Data_Value'].agg(['median','mean']).plot(kind='bar')
plt.xlabel('Stratification1')
plt.ylabel('Data Value')
plt.title('Data Value by Stratification1')
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby('Data_Value_Type')['Data_Value'].agg(['median','mean']).plot(kind='bar')
plt.xlabel('Data Value Type')
plt.ylabel('Data Value')
plt.title('Data Value by Data Value Type')
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby('Topic')['Data_Value'].agg(['median','mean']).plot(kind='bar')
plt.xlabel('Topic')
plt.ylabel('Data Value')
plt.title('Data Value by Topic')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.pie(df['Data_Value_Type'].value_counts(), labels=df['Data_Value_Type'].value_counts().index, autopct='%1.1f%%')
plt.title('Data Value Type Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.pie(df['Stratification1'].value_counts(), labels=df['Stratification1'].value_counts().index, autopct='%1.1f%%')
plt.title('Stratification1 Distribution')
plt.show()
```
