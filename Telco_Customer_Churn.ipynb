```python
import pandas as pd
```

```python
pd.set_option('display.max_columns',None)
```

```python
df=pd.read_csv('/content/WA_Fn-UseC_-Telco-Customer-Churn.csv')
df.head(1)
```

```python
df.drop(['customerID'],axis=1,inplace=True)
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
X=df.drop('Churn',axis=1)
y=df['Churn'].copy()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
```

```python
num_features=X_train.select_dtypes(include='number').columns
cat_features=X_train.select_dtypes(exclude='number').columns
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
```

```python
cat_pipeline=Pipeline([('impute',SimpleImputer(strategy='most_frequent')),
                      ('encode',OneHotEncoder(sparse_output=False,handle_unknown='ignore'))])
num_pipeline=Pipeline([('impute',SimpleImputer(strategy='median')),
 ('scale',StandardScaler())])
```

```python
transformer=ColumnTransformer([('cat_pipeline',cat_pipeline,cat_features),('num_pipeline',num_pipeline,num_features)],remainder='passthrough')
```

```python
X_train_transformed  = transformer.fit_transform(X_train)
```

```python
columns=transformer.get_feature_names_out()
```

```python
X_train_transformed_df=pd.DataFrame(X_train_transformed,columns=columns)
X_train_transformed_df
```

```python
X_test_transformed=transformer.transform(X_test)
```

```python
X_test_transformed_df=pd.DataFrame(X_test_transformed,columns=columns)
X_test_transformed_df
```

```python
rfc=RandomForestClassifier()
```

```python
rfc.fit(X_train_transformed_df,y_train)
```

```python
rfc.score(X_test_transformed_df,y_train)
rfc.score(X_test_transformed_df,y_test)
```
