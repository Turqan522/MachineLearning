```python
X = df.drop(columns=['Name', 'Platform', 'Genre', 'Publisher', 'JP_Sales'])
```

```python
y=df['Genre']
y
```

```python
model=DecisionTreeClassifier()
model.fit(X,y)
```

```python
df
```

```python
prediction = model.predict([[1, 2006, 41.49, 29.02, 3.77, 82.74]])
prediction
```

```python
X_train,X_test,y_train,y_test =train_test_split(X,y,test_size=0.2)
```

```python
model.fit(X_train,y_train)
```

```python
prediction=model.predict(X_test)
prediction
```

```python
score=accuracy_score(y_test,prediction)
score
```
