```python
import pandas as pd
```

```python
df =pd.read_csv('london_merged.csv')
df
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df['timestamp'],df['cnt'])
```

```python
df['timestamp']=pd.to_datetime(df['timestamp'])
```

```python
df['year'] = df['timestamp'].dt.year
df['month'] = df['timestamp'].dt.month
df['day'] = df['timestamp'].dt.day
df['hour'] = df['timestamp'].dt.hour
```

```python
df
```

```python
import seaborn as sns
```

```python
# Yalnız rəqəmsal sütunları seçib korelyasiyanı hesablayırıq
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_df.corr()

# Səliqəli bir heatmap vizuallaşdırması yaradırıq
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Korelyasiya Matrisi', fontsize=16)
plt.show()
```

```python
df_weekday=df[df['is_weekend']==0]
df_weekday=df[df['is_weekend']==1]
```

```python
plt.figure(figsize=(12,5))
plt.tight_layout()
plt.subplot(121)
plt.title("Hourly usage on weekdays")
df_weekday.groupby(by='hour')['cnt'].sum().plot(kind='bar')
plt.subplot(122)
plt.title("Hourly usage on weekends")
df_weekday.groupby(by='hour')['cnt'].sum().plot(kind='bar')
```
