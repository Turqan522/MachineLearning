```python
df.drop(columns=['id','Gender','City','Job Satisfaction'],inplace=True)
df
```

```python
df['Have you ever had suicidal thoughts ?'].head(10000)
```

```python
df["Financial Stress"] = df["Financial Stress"].astype(str)
```

```python
df.isna().sum()
```

```python
print((df["Financial Stress"] == "nan").sum())
```

```python
df['Financial Stress']=pd.to_numeric(df['Financial Stress'],errors='coerce')

df['Financial Stress']=df['Financial Stress'].fillna(df['Financial Stress'].mean())
```

```python
df.isna().sum()
```

```python
df['Age'] = df['Age'].astype(str).str.split('.').str[0]
df['Age']
```

```python
print(df[df["Academic Pressure"] == 0])
```

```python
df=df[df['Academic Pressure']>0]
```

```python
print((df['CGPA']==0).sum())
```

```python
print((df['Study Satisfaction']==0).sum())
```

```python
df=df[df['CGPA']>0]
```

```python
df=df[df['Study Satisfaction']>0]
```

```python
print((df['CGPA']==0).sum())
print((df['Study Satisfaction']==0).sum())
```

```python
df.drop(columns=['Work Pressure'],inplace=True)
```

```python
df
```

```python
print(df["Financial Stress"].min())
```

```python
df.info()
```

```python
df['Age']=pd.to_numeric(df['Age'])
```

```python
df.drop(columns=['Profession'],inplace=True)
```

```python
df.info()
```

```python
for c in ["Have you ever had suicidal thoughts ?", "Family History of Mental Illness"]:
    df[c] = df[c].map({"Yes": 1, "No": 0})
```

```python
df["Sleep Duration"] = df["Sleep Duration"].map({"Less than 5 hours": 0, "5-6 hours": 1, "7-8 hours": 2, "More than 8 hours": 3})
```

```python
df.info()
```

```python
df
```

```python
df.groupby('Dietary Habits')['Academic Pressure'].median()
```

```python
df.groupby('Degree')['CGPA'].agg(['median', 'mean'])
```

```python
df.groupby('Work/Study Hours')['Sleep Duration'].agg(['median','mean'])
```

```python
df.groupby('Age')['CGPA'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 6))
df.groupby('Age')['CGPA'].mean().sort_values(ascending=False).plot(kind='bar')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df.groupby('Work/Study Hours')['Sleep Duration'].agg(['median','mean']).plot(kind='barh')
plt.show()
plt.xlabel('$x')
plt.ylabel('$y')
plt.grid()
```

```python
df.groupby('Degree')['CGPA'].agg(['median', 'mean']).plot(kind='bar')
plt.show()
```

```python
import seaborn as sns

corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```
