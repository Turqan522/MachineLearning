```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
numeric_cols=['Age','Monthly_Income','AI_Usage_Hours_Per_Day','Monthly_AI_Cost','Productivity_Score','Accuracy_Rating','Satisfaction_Score','Time_Saved_Hours_Per_Week','Work_or_Study_Hours_Per_Day']
for col in numeric_cols:
  df[col]=pd.to_numeric(df[col],errors='coerce')
```

```python
for col in numeric_cols:
  df[col]=df[col].fillna(df[col].median())
```

```python
categoric_cols=['Gender','Country','User_Type','Education_Level','Profession','AI_Tool','AI_Purpose','Tasks_Performed','Would_Recommend']
for col in categoric_cols:
  df[col]=df[col].fillna('unknown')
```

```python
plt.figure(figsize=(7,5))

plt.scatter(df['AI_Usage_Hours_Per_Day'], df['Productivity_Score'])

plt.xlabel('AI Usage Hours Per Day')
plt.ylabel('Productivity Score')
plt.title('AI Usage vs Productivity')
plt.grid()
plt.show()
```
