```python
df.drop(columns ='phone',inplace=True)
```

```python
df.drop(columns=['url','address','name'],inplace=True)
```

```python
df['rate'] = pd.to_numeric(df['rate'].astype(str).str.split('/').str[0], errors='coerce')
df['rate'] = df['rate'].fillna(df['rate'].median())
df[['rate']].head()
```

```python
df
```

```python
df['approx_cost(for two people)']=pd.to_numeric(df['approx_cost(for two people)'].astype(str).str.replace(',', '').str.strip(), errors='coerce')
df['approx_cost(for two people)']=df['approx_cost(for two people)'].fillna(df['approx_cost(for two people)'].median())
df[['approx_cost(for two people)']].head()
```

```python
df.isna().sum()
```

```python
df.head(1)
```

```python
df['location']=df['location'].fillna("unknown")
```

```python
df['rest_type']=df['rest_type'].fillna("unknown")
```

```python
df['dish_liked']=df['dish_liked'].fillna('None')
```

```python
df['cuisines']=df['cuisines'].fillna('unknown')
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['rate'],df['votes'])
plt.xlabel('rate')
plt.ylabel('votes')
plt.title('rate vs votes')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['approx_cost(for two people)'],df['votes'])
plt.xlabel('approx_cost(for two people)')
plt.ylabel('votes')
plt.title('approx_cost(for two people) vs votes')
plt.grid()
plt.show()
```

```python
listed=df.groupby('listed_in(city)')['votes'].median()
plt.figure(figsize=(10,5))
listed.plot(kind='bar')
plt.xlabel('listed_in(city)')
plt.ylabel('votes')
plt.title('listed_in(city) vs votes')
plt.show()
```

```python
plt.figure(figsize=(10,5))
df.groupby('online_order')['votes'].median().plot(kind='bar')
plt.xlabel('online_order')
plt.ylabel('votes')
plt.title('online_order vs votes')
plt.show()
#
```

```python
plt.figure(figsize=(10,5))
df.groupby('book_table')['votes'].median().plot(kind='bar')
plt.xlabel('book_table')
plt.ylabel('votes')
plt.title('book_table vs votes')
plt.show()
```

```python
import seaborn as sns
# yalnız kateqorik sütunları seç
cat_df = df.select_dtypes(include=['object', 'category']).copy()

for col in cat_df.columns:
    cat_df[col] = cat_df[col].astype('category').cat.codes

correlation = cat_df.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap (Categorical)')
plt.show()
```

```python
import seaborn as sns
correlation=df.corr(numeric_only=True)
plt.figure(figsize=(10,5))
sns.heatmap(correlation,annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```
