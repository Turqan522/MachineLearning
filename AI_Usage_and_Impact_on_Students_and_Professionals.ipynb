```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
numeric_columns = [
    'Age',
    'Monthly_Income',
    'AI_Usage_Hours_Per_Day',
    'Tasks_Performed',
    'Monthly_AI_Cost',
    'Productivity_Score',
    'Accuracy_Rating',
    'Satisfaction_Score',
    'Time_Saved_Hours_Per_Week',
    'Work_or_Study_Hours_Per_Day'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

*niyə* median mean yox

```python
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = [
    'Gender',
    'Country',
    'Education_Level',
    'Profession',
    'AI_Tool',
    'AI_Purpose',
    'Would_Recommend'
]

for col in categorical_columns:
    df[col] = df[col].fillna('Unknown')
```

```python
df.groupby('Gender')['Productivity_Score'].mean()
```

```python
df.mean(numeric_only=True).plot(
    kind='bar',
    figsize=(12, 5),
    title='Average Values of Numerical Features'
)

plt.ylabel('Mean Value')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.groupby('Education_Level')['Satisfaction_Score'].mean().sort_values(ascending=False)
```

```python
df.groupby('AI_Tool')['Productivity_Score'].agg(['mean', 'count', 'std']).sort_values(by='mean', ascending=False)
```

```python
df.groupby('AI_Purpose')['Time_Saved_Hours_Per_Week'].mean()
```

```python
gender_productivity = df.groupby('Gender')['Productivity_Score'].mean()

plt.figure(figsize=(8,5))
gender_productivity.plot(kind='bar')

plt.title('Average Productivity by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Productivity Score')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

```python
tool_productivity = df.groupby('AI_Tool')['Productivity_Score'].mean().sort_values()

plt.figure(figsize=(10,6))
tool_productivity.plot(kind='barh')

plt.title('Productivity Score by AI Tool')
plt.xlabel('Average Productivity Score')
plt.ylabel('AI Tool')
plt.grid(axis='x')
plt.show()
```

```python
plt.figure(figsize=(9,6))

plt.scatter(
    df['AI_Usage_Hours_Per_Day'],
    df['Productivity_Score'],

)

plt.title('AI Usage Hours vs Productivity')
plt.xlabel('AI Usage Hours Per Day')
plt.ylabel('Productivity Score')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(9,6))

plt.scatter(
    df['Satisfaction_Score'],
    df['Productivity_Score'],
)

plt.title('Satisfaction vs Productivity')
plt.xlabel('Satisfaction Score')
plt.ylabel('Productivity Score')
plt.grid()
plt.show()
```
