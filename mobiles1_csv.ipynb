```python
df['battery']=[a.replace('mAh Battery','') for a in df['battery']]
df['li-ion']=(df['battery'].str.find("Li-Ion"))
```

```python
df.drop(columns=['battery'],inplace=True)
```

```python
df
```

```python
df['brand'] = [a.split()[0] for a in df['name']]
df['colour'] = [a.split()[3] for a in df['name']]
df['model'] = [a.split()[1] for a in df['name']]
```

```python
df
```

```python
df['processor_name']=[a.split()[0] for a in df['processor']]
df.drop(columns=['processor'],inplace=True)
```

```python
df['rear_in_mp']=[a.split('|')[0].replace('Rear Camera','').replace('MP','') for a in df['camera']]
df['front_in_mp']=[a.split('|')[-1].replace('Front Camera','').replace('MP','') for a in df['camera']]
df.drop(columns=['camera'],inplace=True)
```

```python
df['reviews']=[a.replace('Reviews','').replace(',','') for a in df['reviews']]
```

```python
df
```

```python
df.groupby('memory')['rating'].agg(['mean','count'])
```

```python
df.groupby('name')['price'].agg(['median','mean'])
```

```python
df.groupby('reviews')['battery_value'].median()
```

```python
df.groupby('warranty')['price'].median()
```

```python
df['brand'].value_counts().plot(kind='barh',figsize=(15,15))
```

```python
plt.figure(figsize=(10,6))
df.groupby('warranty')['price'].median().plot(kind='bar')
plt.xlabel('warranty')
plt.ylabel('price')
plt.title('warranty vs price')
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby('price')['rating'].plot(kind='bar')
plt.xlabel('price')
plt.ylabel('rating')
plt.title('price vs rating')
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby('brand')['price'].median().plot(kind='bar')
plt.xlabel('brand')
plt.ylabel('price')
plt.title('brand vs price')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['colour'],df['price'])
plt.xlabel('colour')
plt.ylabel('price')
plt.title('colour vs price')
plt.show()
```
